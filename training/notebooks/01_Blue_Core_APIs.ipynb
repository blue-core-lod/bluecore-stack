# Interacting with Blue Core API
The Blue Core API offers endpoints for retrieving (GET), creating (POST), updating (PUT), and deleting (DELETE) BIBFRAME resources:
- [Hubs](https://id.loc.gov/ontologies/bibframe.html#c_Hub)
- [Works](https://id.loc.gov/ontologies/bibframe.html#c_Work)
- [Instances](https://id.loc.gov/ontologies/bibframe.html#c_Instance).

The OpenAPI view of the Blue Core API is available at https://dev.bcld.info/api/docs.

The GET responses do not require authenication to the API. The other operations for resources do require a valid token.

## Retrieving a Token from Keycloak
The `get_token` function below demostrates how to authenticate and then retrieve a JWT from Keycloak for use in subsequent calls to the Blue Core API that require authorization. This function is also available in the `helpers.py` Python module for use in other notebooks.

```python
import os

def get_token(url=None, username="developer", password="123456"):
    # Inside JupyterHub, KEYCLOAK_INTERNAL_URL points straight at Keycloak on
    # the shared Docker network (bypassing nginx, whose external listen
    # port/scheme varies by deployment). Outside JupyterHub, fall back to the
    # locally published nginx route.
    if url is None:
        url = os.environ.get("KEYCLOAK_INTERNAL_URL", "http://localhost/keycloak")
    realm = os.environ.get("KEYCLOAK_REALM", "bluecore")
    keycloak_token = httpx.post(f"{url.rstrip('/')}/realms/{realm}/protocol/openid-connect/token",
                      data={
                          "client_id": "bluecore_api",
                          "username": username,
                          "password": password,
                          "grant_type": "password",
                      })
    
    keycloak_token.raise_for_status()
    return keycloak_token.json().get('access_token')
```
