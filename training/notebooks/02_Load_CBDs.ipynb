# Loading LOC CBD files
As explained in the README, there are scripts in the Blue Core Stack where you can load CBD files from either
a URL or you can load CBD files directly to the API. Below is an example of ingesting a local CBD file using the Blue Core API.

```python
import httpx

from helpers import get_token
```

## Search for a BIBFRAME Instance
At https://id.loc.gov/resources/instances.html, search for an BIBFRAME Instance on the topic of your choice.

## Retrieve CBD File with HTTPX
After finding an Instance, download the CBD with  adding `.cbd.rdf` to the URL i.e.
`https://id.loc.gov/resources/instances/23236867.cbd.rdf`

```python
# Replace with the specific Instance URL CBD RDF URL
instance_result = httpx.get("https://id.loc.gov/resources/instances/23236867.cbd.rdf") 
```

Retrieve the RDF XML from the `instance_result` and display the XML:

```python
instance_rdf = instance_result.text
print(instance_rdf)
```

## Upload the CBD XML to the Blue Core API

```python
post_result = httpx.post("http://localhost/api/batches/upload/",
                         headers={
                             "Authorization": f"Bearer {get_token()}"
                         },
                         follow_redirects=True,
                         files={"file": instance_rdf})
```

## Observe `resource_loader` DAG in Airflow
First log into the Blue Core Workflows Airflow environment at `http://localhost/workflows` and 
then go to the `resource_loader` DAG at http://localhost/workflows/dags/resource_loader to see the CBD
file being loaded.


## Search for Works and Instances
After the DAG run is completed, go to `http://localhost/api/search` and search for the Works and Instances
from the CBD file you uploaded.
